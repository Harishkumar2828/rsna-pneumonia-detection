```python
import pandas as pd

# Load the annotation CSV
labels = pd.read_csv("data/raw/stage_2_train_labels.csv")

# Display first 5 rows
print("FIRST 5 ROWS:")
display(labels.head())

# Display dataset dimensions
print("\nDATASET SHAPE:")
print(labels.shape)

# Display column names
print("\nCOLUMN NAMES:")
print(labels.columns.tolist())
```

```python
# Count records in each target class
print("TARGET DISTRIBUTION:")
print(labels["Target"].value_counts())

# Calculate percentages
print("\nTARGET PERCENTAGES:")
print(labels["Target"].value_counts(normalize=True) * 100)

# Check missing values in each column
print("\nMISSING VALUES:")
print(labels.isnull().sum())
```

```python
# Count unique patients in each class
print("UNIQUE PATIENTS BY CLASS:")
print(labels.groupby("Target")["patientId"].nunique())

# Count bounding boxes per positive patient
positive_patients = labels[labels["Target"] == 1]

boxes_per_patient = positive_patients.groupby("patientId").size()

print("\nPATIENTS WITH DIFFERENT NUMBERS OF BOXES:")
print(boxes_per_patient.value_counts().sort_index())

print("\nTOTAL POSITIVE PATIENTS:")
print(boxes_per_patient.nunique() if False else len(boxes_per_patient))
```

```python
# Find the ID with the maximum number of bounding boxes
sample_id = boxes_per_patient.idxmax()

# Display its annotations
sample_annotations = labels[
    labels["patientId"] == sample_id
]

print("SAMPLE PATIENT ID:")
print(sample_id)

print("\nNUMBER OF BOUNDING BOXES:")
print(len(sample_annotations))

print("\nANNOTATION DETAILS:")
display(sample_annotations)
```

```python
import pydicom
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Locate the original DICOM image
image_path = f"data/processed/train_images/{sample_id}.dcm"

# Read DICOM image
dicom_data = pydicom.dcmread(image_path)
image = dicom_data.pixel_array

# Display image
fig, ax = plt.subplots(figsize=(10, 10))

ax.imshow(image, cmap="gray")

# Draw all bounding boxes
for _, row in sample_annotations.iterrows():

    rectangle = patches.Rectangle(
        (row["x"], row["y"]),
        row["width"],
        row["height"],
        linewidth=2,
        edgecolor="red",
        facecolor="none"
    )

    ax.add_patch(rectangle)

ax.set_title("RSNA Pneumonia Bounding Box Annotations")
ax.axis("off")

plt.show()
```
